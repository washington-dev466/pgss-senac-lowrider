Faça um algoritmo para ler três números e imprimir se estes podem ou não formar um triângulo.

1. Informar primeiro número;
2. Informar segundo número;
3. Informar terceiro número;
4. Verificar se o primeiro, for menor que a soma dos outros dois;
5. Verificar se o segundo, for menor que a soma dos outros dois;
6. Verificar se o terceiro, for menor que a soma dos outros dois;
7. Após verificar e for verdadeiro as 3 verificações anteriores, informar que sim, podem formar um triângulo;
6. Se não, informar que não podem formar um triângulo.

```python
#Faça um algoritmo para ler três números e imprimir se estes podem ou não formar um triângulo.
H = int(input('Informe um número: '))
I = int(input('Informe outro número: '))
J = int(input('Informe mais um número: '))
print(type(H))

if H < I + J and I < H + J and J < H + I:
  print(f'Os números {H}, {I} e {J} podem formar um triângulo.')

else:
  print('Os números não podem formar um triângulo.')
```

Faça um algoritmo para ler três números e imprimir se estes podem ou não formar um triângulo.

1. Informar primeiro número;
2. Informar segundo número;
3. Informar terceiro número;
4. Verificar se o primeiro, for menor que a soma dos outros dois;
5. Verificar se o segundo, for menor que a soma dos outros dois;
6. Verificar se o terceiro, for menor que a soma dos outros dois;
7. Após verificar e for verdadeiro as 3 verificações anteriores, informar que sim, podem formar um triângulo.
8. Se não, informar que não podem formar um triângulo.
9. Caso seja possível formar um triângulo, verificar as seguintes condições;
10. Se o primeiro for igual ao segundo e o segundo for igual ao terceiro, informar que é equilátero;
11. Se o primeiro for igual ao segundo ou o segundo for igual ao terceiro, informar que é isósceles;
12. Se não, informar que é escaleno.

```python
#Faça um algoritmo para ler três números e se estes poderem formar um triângulo dizer se o triângulo é “EQUILÁTERO”, “ISÓSCELES” OU “ESCALENO”.
K = int(input('Informe um número: '))
L = int(input('Informe outro número: '))
M = int(input('Informe mais um número: '))
print(type(K))

if (K + L > M) and (K + M > L) and (L + M > K):
  print("Os valores podem formar um triângulo:")

  if K == L and L == M:
    print("EQUILÁTERO")
  elif K == L or K == M or L == M:
    print("ISÓSCELES")
  else:
    print("ESCALENO")
else:
  print("Os valores NÃO podem formar um triângulo.")
```

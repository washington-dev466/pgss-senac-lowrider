```python
#Informar a soma, média simples e produto dos números indicados
n1 = int(input('Informe um número: '))
n2 = int(input('Informe outro número: '))
n3 = int(input('Informe mais um número: '))
print(type(n1))
soma = n1 + n2 + n3
média = soma / 3
produto = n1 * n2 * n3
print(f'O resultado é: {soma}, {média} e {produto} ')
```

```python
#Verificar se os números informados, um é maior do que o outro ou se são iguais
n1 = int(input('Informe um número: '))
n2 = int(input('Informe outro número: '))
print(type(n1))

if n1 < n2:
  print(f'{n1} é menor que {n2}')

elif n1 > n2:
  print(f'{n1} é igual a {n2}')

else:
  print(f'Os números são iguais')
```

```python
#Faça um algoritmo para ler três números e imprimir se estes podem ou não formar um triângulo.
H = int(input('Informe um número: '))
I = int(input('Informe outro número: '))
J = int(input('Informe mais um número: '))
print(type(H))

if H < I + J and I < H + J and J < H + I:
  print(f'Os números {H}, {I} e {J} podem formar um triângulo.')

else:
  print('Os números não podem formar um triângulo.')
```

```python
#Faça um algoritmo para ler três números e se estes poderem formar um triângulo dizer se o triângulo é “EQUILÁTERO”, “ISÓSCELES” OU “ESCALENO”.
K = int(input('Informe um número: '))
L = int(input('Informe outro número: '))
M = int(input('Informe mais um número: '))
print(type(K))

if (K + L > M) and (K + M > L) and (L + M > K):
  print("Os valores podem formar um triângulo:")

  if K == L and L == M:
    print("EQUILÁTERO")
  elif K == L or K == M or L == M:
    print("ISÓSCELES")
  else:
    print("ESCALENO")
else:
  print("Os valores NÃO podem formar um triângulo.")
```

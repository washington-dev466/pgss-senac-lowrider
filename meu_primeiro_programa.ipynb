# Exercícios de Programação

1. Faça um algoritmo para ler dois números e imprimir a soma dos números lidos.



## Algorítmo

1. Leia o primeiro número e guarde uma variável chamada n1
2. Leia o segundo número e guarde em outra variável chamada n2
3. Faça a soma das duas varáveis e guarde o resultado em outra variável chamada n3
4. Mostre o resultado

```python
n1 = 5
n2 = 6
soma = n1 + n2
print(f'O resultado da soma entre n1 e n2 é {soma}.')
```

```python
n1 = int(input('Informe um número: '))
n2 = int(input('Informe outro número: '))
print(type(n1))
soma = n1 + n2
print(f'O resultado é: {soma}')
```

```python
n1 = int(input('Informe um número: '))
n2 = int(input('Informe outro número: '))
n3 = int(input('Informe mais um número: '))
print(type(n1))
soma = n1 + n2 + n3
média = soma / 3
produto = n1 * n2 * n3
print(f'O resultado é: {soma}, {média} e {produto} ')
```

```python
n1 = int(input('Informe um número: '))
n2 = int(input('Informe outro número: '))
print(type(n1))

if n1 < n2:
  print(f'{n1} é menor que {n2}')

elif n1 > n2:
  print(f'{n1} é igual a {n2}')

else:
  print(f'Os números são iguais')
```

```python
n1 = int(input('Informe um número: '))
print(type(n1))

if n1 % 2 == 0:
  print(f'{n1} é par')

else:
  print(f'{n1} é ímpar')
```
